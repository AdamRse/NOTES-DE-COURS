# Numpy

## List vs Array

- liste Python, c'est comme une boite a outils. Chaque element est independant, donc les ranger prend du temps.
- Un array Numpy (ndarray), c'est un tournevis electrique. Operations se font ensemble,
d'un coup, en masse.

## Creation d'arrays et attributs

```python
l = [1,2,3,4,5,6,7,8,9]
a = np.array(l)

print(a)
print(type(a))
print(a.shape)
```

```python
liste = [1,2,3,4,5]
a = np.array([1,2,3,4,5])

#operation sur la liste
print([x*2 for x in liste])

# operation sur array
print(a * 2)
```

## Difference de vitesse

```python
import time
liste = list(range(10_000_000))
a = np.arange(10_000_000)

#liste
t0 = time.perf_counter()
resultat_liste = [x*2 for x in liste]
t1 = time.perf_counter()

#array
t2 = time.perf_counter()
resultat_array = a * 2
t3 = time.perf_counter()

print(f"Liste Python : {(t1 - t0)*1000:.2f} ms")
print(f"NumPy        : {(t3 - t2)*1000:.2f} ms")
print(f"Accélération : {(t1 - t0)/(t3 - t2):.1f}x")
```

## Les fonctions constructeurs

```python
print(f"np.arange(10) = {np.arange(10)}")
print(f"np.arange(2, 10, 2) = {np.arange(2, 10, 2)}")

print(f"np.zeros(5) = {np.zeros(5)}")
print(f"np.ones(5) = {np.ones(5)}")
print(f"np.ones(5) = \n{np.ones((3,4))}")

# points egalement espacés
print(f"np.linspace(0, 100, 5) = {np.linspace(0, 100, 5)}")

#reshape
print(f"{np.arange(12).reshape(3, 4)}")

#
print(np.arange(12).reshape(1,12).shape)
```

## les attributs

```python
a = np.arange(24).reshape(4,6)
print(a)
print(f"Forme : {a.shape}")
print(f"Nombre dimensions : {a.ndim}")
print(f"Type : {a.dtype}")
print(f"Taille : {a.size}")
print(f"Taille element : {a.itemsize}")
print(f"Memoire totale : {a.nbytes}")
```

## Génération aléatoire
